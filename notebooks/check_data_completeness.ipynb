```python
import pandas as pd
import os

# Since you're in /notebooks/, go up one level with ../
correct_path = "../data/processed/all_races_combined_{year}.csv"

print("Current directory:", os.getcwd())
print(f"Looking in: {os.path.abspath('../data/processed/')}")
print(f"Does directory exist? {os.path.exists('../data/processed/')}\n")

# List CSV files if directory exists
if os.path.exists('../data/processed/'):
    print("CSV files found:")
    csv_files = [f for f in os.listdir('../data/processed/') if f.endswith('.csv')]
    for f in csv_files:
        print(f"  - {f}")
    print()

print("="*60)
print("DATA COMPLETENESS CHECK")
print("="*60 + "\n")

expected_races = {
    2018: 21, 2019: 21, 2020: 17,
    2021: 22, 2022: 22, 2023: 22, 2024: 24, 2025: 24
}

for year in range(2018, 2026):
    try:
        file_path = correct_path.format(year=year)
        df = pd.read_csv(file_path)
        num_races = df['GrandPrix'].nunique()
        expected = expected_races.get(year, "Unknown")
        status = "✅" if num_races >= expected else "❌"
        print(f"{status} {year}: {num_races} races (expected: {expected})")
        
        # Show which races
        races = sorted(df['GrandPrix'].unique())
        print(f"   Races: {', '.join(races[:3])}... (showing first 3)")
        print()
    except FileNotFoundError:
        print(f"❌ {year}: File not found\n")
    except Exception as e:
        print(f"❌ {year}: Error - {e}\n")
```
