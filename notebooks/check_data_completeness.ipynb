```python
import pandas as pd

expected_races = {
    2018: 21, 2019: 21, 2020: 17,
    2021: 22, 2022: 22, 2023: 22, 2024: 24, 2025: 24
}

for year in range(2018, 2026):
    try:
        file_path = f"../data/processed/all_races_combined_{year}.csv"
        df = pd.read_csv(file_path)
        num_races = df['GrandPrix'].nunique()
        expected = expected_races.get(year, "Unknown")
        status = "✅" if num_races >= expected else "❌"
        print(f"{status} {year}: {num_races} races (expected: {expected})")
        
        # Show which races
        races = sorted(df['GrandPrix'].unique())
        print(f"   First 3: {', '.join(races[:3])}")
        print()
    except Exception as e:
        print(f"❌ {year}: {e}\n")
```
